## 1.3.1 Why Do We Need Package Managers?

As discussed in Section 1.8, much of Python's functionality comes from packages developed by the scientific and programming communities. These packages must be downloaded, installed, and maintained on our computers.

Python packages are distributed through online repositories, including the [Python Package Index (PyPI)](https://pypi.org/) and [conda-forge](https://conda-forge.org/). PyPI is the primary repository for packages installed using `pip`, while conda-forge is a community-maintained repository of packages installed using `conda`. We will learn more about these tools later in this notebook.

Installing packages, however, is not always straightforward. Many packages depend on other packages, which are called **dependencies**. For example, **[NumPy](https://numpy.org/)** (Numerical Python) provides tools for numerical calculations and working with arrays of data. It is a foundational package used by many other scientific Python packages, including **[Matplotlib](https://matplotlib.org/)**, which creates graphs and visualizations, and **[pandas](https://pandas.pydata.org/)**, which organizes and analyzes tabular data. Both depend on NumPy for numerical operations. Furthermore, different packages may require specific versions of Python or their dependencies. A **package manager** helps us install software, identify dependencies, and find compatible versions.

One major challenge is that **updating one package can break another**. For example, suppose a scientific program requires an older version of NumPy. Installing another program that requires a newer version of NumPy may cause the original program to stop working. A package manager can help identify these conflicts, but it cannot always resolve incompatible requirements within a single Python installation.

Another concern is protecting the operating system itself. Linux distributions such as Ubuntu and Raspberry Pi OS use Python for some system utilities, and modifying their Python installations can interfere with system software. Python's development is guided in part by Python Enhancement Proposals (PEPs), which are documents used to propose, discuss, and standardize changes to Python and its ecosystem. One example is PEP 668, which allows operating systems to designate their Python installations as externally managed, meaning that the operating system's package manager (such as Ubuntu's apt), rather than Python's pip, is responsible for maintaining them. This prevents pip from inadvertently modifying software that other system programs depend on.

The solution is to use **isolated Python environments**, each containing its own Python interpreter and collection of packages. This allows different projects to use different versions of Python and their dependencies without interfering with one another.

In this course, we will use Miniconda, which provides the conda package and environment manager. Conda allows us to install and manage Python and its scientific packages independently of the operating system's Python installation. We will learn to create separate environments for our scientific computing projects, allowing us to install, update, and maintain packages without interfering with other projects or the operating system.

## 1.3.2 Python Package and Environment Management: pip, venv, and Conda

There are several tools for installing Python packages and managing the environments in which they run. Three that you are likely to encounter are `pip`, `venv`, and `conda`. Although their functions overlap, they serve different purposes.

### pip: Installing Python Packages

**pip** is the standard package installer for Python. It is commonly used to download and install packages from the [Python Package Index (PyPI)](https://pypi.org/).

For example, scientific software documentation often includes installation instructions such as:

```bash
pip install numpy
```

This command tells `pip` to install NumPy and any required dependencies into the Python environment associated with that `pip` command.

However, **pip does not create or manage separate Python environments**. It installs packages into an existing environment, which may be the operating system's Python installation or an isolated environment created for a particular project.

Consequently, before running a `pip install` command, it is important to know which Python environment will receive the package.

### venv: Creating Python Virtual Environments

Python includes a built-in module called **venv** for creating isolated Python environments. Each environment has its own package installation directory and uses a Python interpreter associated with that environment.

For example, the command:

```bash
python -m venv .venv
```

creates a virtual environment in a directory named `.venv`. The leading period makes this a hidden directory on Unix-like systems, and `.venv` is a common naming convention rather than a required name.

Once the environment is activated, packages can be installed into it using `pip` without modifying the operating system's Python installation.

Thus, `venv` and `pip` are commonly used together: **venv creates the environment, while pip installs packages into it.**

### conda: Managing Packages and Environments

**Conda** combines package management and environment management into a single tool. It can create isolated environments, install different versions of Python, and manage both Python packages and other software libraries required by scientific applications.

Unlike `pip`, which primarily installs Python packages from PyPI, Conda obtains packages from repositories called **channels**. One widely used channel is [conda-forge](https://conda-forge.org/), a community-maintained repository containing scientific software and its dependencies.

Conda is particularly useful in scientific computing, where packages may depend on compiled libraries and software written in languages other than Python.

In this course, we will use **Conda as our primary package and environment manager**. This provides a consistent approach for installing Python, managing scientific packages, and maintaining separate computing environments.

You will still encounter `pip` installation instructions, and there will be situations where we use `pip` within a Conda environment. However, our general practice will be to look for packages on conda-forge first, using `pip` when necessary.

We will also return to `venv` later in the course so that you gain experience with both approaches to managing Python environments.

The next section introduces the Conda ecosystem and the differences between Conda, Anaconda, and Miniconda.

## 1.3.3 The Conda Ecosystem

Conda is an open-source package and environment manager widely used in scientific computing. It is developed and maintained through contributions from an international community of developers and organizations. The broader Conda ecosystem includes software distributions, graphical interfaces, and community-maintained package repositories that help scientists install and manage the software needed for their research.

- **[Conda](https://docs.conda.io/):** The package and environment manager used to install software, resolve dependencies, and create isolated computing environments.
- **[Anaconda Distribution](https://www.anaconda.com/docs/getting-started/anaconda/main):** A comprehensive Python distribution that includes Conda, Python, and hundreds of scientific and data analysis packages.
- **[Anaconda Navigator](https://www.anaconda.com/products/navigator) and [Anaconda Desktop](https://www.anaconda.com/products/desktop):** Graphical interfaces for exploring and managing packages, creating computing environments, and launching applications such as JupyterLab. Anaconda Desktop is the newer application and adds capabilities for discovering and running AI models locally.
- **[Miniconda](https://www.anaconda.com/docs/getting-started/miniconda/main):** A minimal distribution containing Conda, Python, and the essential software needed to create and manage environments.
- **[conda-forge](https://conda-forge.org/):** A community-maintained repository of software packages that can be installed using Conda.

These components provide complementary ways to access and manage scientific software. Anaconda Distribution offers a ready-to-use collection of scientific tools, while Miniconda provides a smaller starting point from which users can install the software they need. Both can access packages from repositories such as conda-forge.

### Anaconda Distribution and Graphical Interfaces

[Anaconda Distribution](https://www.anaconda.com/docs/getting-started/concepts/anaconda-or-miniconda) provides a ready-to-use scientific computing environment that includes Python, Conda, JupyterLab, and hundreds of preinstalled packages for data analysis, visualization, machine learning, and other applications. This allows users to begin exploring scientific Python without first identifying and installing individual packages.

Anaconda also provides graphical interfaces for managing these resources. **[Anaconda Navigator](https://www.anaconda.com/products/navigator)** allows users to discover and install packages, create and manage environments, and launch applications such as JupyterLab without using terminal commands. Its newer successor, **[Anaconda Desktop](https://www.anaconda.com/products/desktop)**, provides similar capabilities while adding tools for discovering and running artificial intelligence models locally.

These tools offer a convenient way to explore the scientific Python ecosystem, discover new software, and manage computing environments. Although we will use Miniconda and the command-line interface in this course, Anaconda's graphical tools remain valuable resources for students interested in exploring additional scientific applications.

### Miniconda and the Command-Line Interface

In this course, we will install [Miniconda](https://www.anaconda.com/docs/getting-started/concepts/anaconda-or-miniconda) and interact with Conda through the **command-line interface (CLI)**. Once installed, Conda provides a new set of commands that we can execute from our terminal. Just as we use Unix commands such as `cd` and `ls` to navigate the filesystem, we will use commands beginning with `conda` to create environments, install packages, and manage our scientific software.

We are choosing this approach for several reasons:

1. **Smaller initial installation:** We install only the scientific packages we need rather than downloading a large collection of software that may never be used.
2. **Understanding our computing environments:** By creating environments ourselves, we learn how Python versions, scientific packages, and dependencies fit together.
3. **Transferable skills:** Command-line tools can be used on personal computers, remote research servers, and high-performance computing systems where graphical interfaces may not be available.

Our goal is not simply to use scientific software, but to develop the skills needed to **build, understand, and maintain our own scientific computing environments**. As the course progresses, we will learn to install packages from repositories such as conda-forge and PyPI, obtain and work with scientific code shared through GitHub, and configure environments to meet the needs of different scientific projects.

## 1.3.4 Installing Miniconda

We will install Miniconda directly from the terminal, using the operating system environment configured in Notebook 1.2. Windows users will install it inside Ubuntu through Windows Subsystem for Linux (WSL), while macOS users will install it directly in macOS.

The installation involves four basic steps:

1. Download the appropriate Miniconda installer.
2. Run the installer and follow its instructions.
3. Configure the terminal to recognize Conda commands.
4. Verify that Conda was installed successfully.

The official [Miniconda installation documentation](https://www.anaconda.com/docs/getting-started/miniconda/install) provides additional installation instructions and troubleshooting information.

### Windows Users: Installing Miniconda in WSL Ubuntu

If you are using Windows, open your Ubuntu terminal. We will install the Linux version of Miniconda inside WSL, not the Windows version. This is important because Ubuntu and Windows are separate operating environments, even though they run on the same computer. Installing Miniconda in Ubuntu allows us to use Linux-based Python environments and tools directly from the Ubuntu terminal.

**Step 1: Navigate to your home directory**

Enter:

```bash
cd ~
```

The `~` represents your home directory. We will download the installer here and allow Miniconda to create its installation directory inside your home directory.

**Step 2: Download the Miniconda installer**

For most Windows computers running Ubuntu through WSL, the appropriate installer is the Linux x86-64 version:

```bash
wget https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh
```

The `wget` command downloads a file from the internet. The filename identifies the operating system (Linux), processor architecture (`x86_64`), and file type (`.sh`, a shell script).

If your computer uses an ARM processor, you may need a different installer. You can check the processor architecture with `uname -m`. The standard installer above is appropriate when the result is `x86_64`.

**Step 3: Run the installer**

Once the download is complete, enter:

```bash
bash Miniconda3-latest-Linux-x86_64.sh
```

This tells the Bash shell to execute the installation script.

The installer will guide you through several steps:

- Press **Enter** to review the license agreement. Use Enter or Space to scroll through it, and enter `yes` if you agree to its terms.
- Accept the default installation location, which is typically `~/miniconda3`.
- Allow the installation to complete.
- When asked whether to initialize Conda by modifying your shell profile, enter `yes`.

Initializing Conda allows your terminal to recognize Conda commands when you open a new shell.

**Do not use `sudo` to install Miniconda.** We are installing it in your own home directory, where administrator privileges are not required.

### macOS Users: Installing Miniconda

Mac users can install Miniconda directly in macOS without installing Ubuntu.The procedure is similar to Linux, but the installer must match your Mac's processor architecture.

**Step 1: Open the Terminal application**

Open Terminal and navigate to your home directory:

```bash
cd ~
```

**Step 2: Identify your processor architecture**

Enter:

```bash
uname -m
```

For a Mac with Apple Silicon (M-series processors), the result is typically:

```text
arm64
```

For an older Intel-based Mac, the result is typically:

```text
x86_64
```

**Step 3: Download the appropriate installer**

For a Mac with Apple Silicon, enter:

```bash
curl -O https://repo.anaconda.com/miniconda/Miniconda3-latest-MacOSX-arm64.sh
```

The `curl` command downloads the installer, and the `-O` option saves it using its original filename.

For Intel-based Macs, consult the [official Miniconda installer archive](https://repo.anaconda.com/miniconda/) for the appropriate Intel-compatible installer. Anaconda has ended new Miniconda releases for Intel macOS, so the most recent compatible installer may be older than the Apple Silicon version.

**Step 4: Run the installer**

For Apple Silicon Macs, enter:

```bash
bash Miniconda3-latest-MacOSX-arm64.sh
```

Follow the installation prompts, accept the default installation location, and allow Conda to initialize your shell when prompted. Most modern macOS installations use the **Z shell (`zsh`)** as their default terminal shell. The installer should configure Conda for the shell you are using.

### Verifying the Installation

After the installation finishes, **close your terminal and open a new one**. This allows the shell to load the configuration changes made during installation.

Enter:

```bash
conda --version
```

If Miniconda was installed successfully, the terminal should display the installed Conda version, for example:

```text
conda 25.x.x
```

The exact version number may differ depending on when you installed Miniconda.

You can also enter:

```bash
conda info
```

This displays information about your Conda installation, including its version, installation location, active environment, and configured package repositories. Look for a line identifying the location of the **base environment**. For an installation in your home directory, it will resemble:

```text
base environment : /home/student/miniconda3
```

On macOS, the corresponding path will typically begin with `/Users/` rather than `/home/`. You may also notice that your terminal prompt now begins with `(base)`, for example:

```text
(base) student@computer:~$
```

This indicates that Conda's default environment, called `base`, is active. We will explore what this means in the next section.

### What Have We Installed?

We have installed Miniconda and configured our terminal to recognize Conda commands. Miniconda provides a Python interpreter, the Conda package and environment manager, and the supporting software needed for Conda to operate. It does not include the extensive collection of scientific packages provided by the full Anaconda Distribution.

We now have the tools needed to begin creating and managing our own Python environments.

In the next section, we will examine the Conda `base` environment and explain why we generally avoid installing our scientific project packages directly into it.

## 1.3.5 Understanding the Conda Base Environment

When we installed Miniconda, it created a special environment called **`base`**, which contains Conda itself, a Python interpreter, and the supporting software needed for Conda to operate. The `base` environment is primarily used to maintain Conda and manage other environments. Although we can install scientific packages directly into `base`, doing so can introduce dependency conflicts and make Conda more difficult to maintain. **Our practice will be to keep `base` relatively clean and create separate environments for our scientific projects, including the Jupyter Lab environment.**

### Activating and Deactivating Environments

You may have noticed that your terminal prompt changed after installing Miniconda. For example, it might now look like:

```text
(base) student@computer:~$
```

The `(base)` indicates that the Conda `base` environment is currently **active**. An active environment determines which Python interpreter and installed software are normally used when we execute commands in the terminal.

Let's explore how to activate and deactivate an environment.

**Step 1: Deactivate the base environment**

If your terminal prompt begins with `(base)`, enter:

```bash
conda deactivate
```

You should notice that `(base)` disappears from your terminal prompt:

```text
student@computer:~$
```

This indicates that `base` is no longer active. Your terminal is now using the software available through your normal shell configuration, which may include the operating system's Python installation.

**Step 2: Activate the base environment**

Enter:

```bash
conda activate base
```

Your terminal prompt should change to something similar to:

```text
(base) student@computer:~$
```

You have now activated the Conda `base` environment.

Notice that we did not open a new application or terminal. Instead, Conda modified our current shell environment so that commands such as `python` use the Python interpreter and packages associated with `base`environment.

**Step 3: Deactivate base again**

Enter:

```bash
conda deactivate
```

The `(base)` prefix should disappear.

Importantly, **deactivating an environment does not delete it or remove any installed packages**. It simply changes which environment is active in the current terminal.

### Controlling Automatic Activation of Base

By default, Miniconda may configure your terminal to automatically activate `base` whenever you open a new terminal session. Although convenient, we will configure Conda so that `base` is not automatically activated. This helps us distinguish between the operating system's Python installation and the Conda environments we will create for our projects.

Enter:

```bash
conda config --set auto_activate_base false
```

This changes Conda's configuration so that new terminal sessions do not automatically activate `base`.

**Close your terminal and open a new one.**

Your terminal prompt should no longer begin with `(base)`. However, Conda is still available. You can verify this by entering:

```bash
conda --version
```

You should see the Conda version number, just as you did after installing Miniconda. If you ever need to activate `base`, simply enter:

```bash
conda activate base
```

You can leave it again with:

```bash
conda deactivate
```

### What Have We Learned?

We have learned that Conda environments can be activated and deactivated independently of the terminal itself. Activating an environment makes its Python interpreter and installed packages available to our current terminal session. We have also learned that **Conda does not need to have `base` activated to manage environments**. We can use Conda commands to create, inspect, and manage environments without working directly inside `base`.

In the next section, we will begin exploring Conda commands and learn how to create our own Python environments.


## 1.3.6 Creating and Exploring Conda Environments

Now that we have installed Miniconda and learned how to activate and deactivate environments, we are ready to create our first scientific Python environment.

We will name this environment **`py4sci`**, and it will contain the Python interpreter and many of the scientific packages we will use throughout this course.

After creating the environment, we will explore how Conda organizes and manages environments using commands entered in the terminal.

### 1.3.6.1 Creating Our First Conda Environment

We are now ready to create our first Conda environment. We will name it **`myenv`** and install Python along with several packages that we will use for scientific computing.

Our environment will contain:

- **Python:** The interpreter that executes our Python code.
- **NumPy:** A package for numerical calculations and working with arrays of scientific data.
- **Matplotlib:** A package for creating graphs and scientific visualizations.
- **ipykernel:** Software that allows a Python environment to execute code from Jupyter notebooks.

**Step 1: Create the environment**

Open your terminal and enter the following command:

```bash
conda create -n myenv -c conda-forge --override-channels python=3.12 numpy matplotlib ipykernel
```

Let's examine the components of this command:

- `conda create` tells Conda to create a new environment.
- `-n myenv` specifies the name of our environment.
- `-c conda-forge` tells Conda to obtain packages from the conda-forge repository.
- `--override-channels` instructs Conda to use only the specified channel for this installation, rather than also searching other configured channels.
- `python=3.12` specifies the Python version we want to install.
- `numpy matplotlib ipykernel` identifies the additional packages we want in the environment.

Conda will examine the requested packages, determine their dependencies, and identify compatible versions to install. You will notice that Conda may list many more packages than the three we requested. These additional packages are dependencies required by Python or the scientific packages we are installing.

After reviewing the proposed installation, enter `y` and press Enter to proceed. Depending on your computer and internet connection, the installation may take several minutes.

**Step 2: Activate the environment**

Once Conda finishes installing the packages, enter:

```bash
conda activate myenv
```

Your terminal prompt should now begin with `(myenv)`, indicating that the environment is active.

For example:

```text
(myenv) student@computer:~$
```

When `myenv` is active, commands such as `python` will use the Python interpreter and packages installed in this environment.

**Step 3: Verify the Python installation**

Enter:

```bash
python --version
```

The output should indicate Python 3.12, although the final version numbers may vary.

Now let's test NumPy by calculating the average of five temperature measurements. Enter the following command directly into the terminal:

```bash
python -c "import numpy as np; print(np.mean([21.3, 22.1, 20.8, 23.0, 21.7]))"
```

The `python -c` option allows us to execute Python instructions directly from the terminal without creating a Python script.

This command imports NumPy, calculates the mean of the five measurements, and prints the result:

```text
21.78
```

If you see this result, you have successfully created a Conda environment and used one of its scientific packages.

**Step 4: Deactivate the environment**

Enter:

```bash
conda deactivate
```

The `(myenv)` prefix should disappear from your terminal prompt. The environment and its installed packages are still on your computer. You can activate `myenv` again whenever you need it by entering `conda activate myenv`.

We have now created our first scientific Python environment. In the next section, we will create a second environment specifically for JupyterLab, which will provide the interactive interface used to work with our scientific notebooks.

### 1.3.6.2 Creating a JupyterLab Environment (`jlab`)

We will now create a second Conda environment named **`jlab`**, which will contain [JupyterLab](https://jupyterlab.readthedocs.io/), the interactive development environment we will use to work with Python notebooks throughout this course.

Unlike `myenv`, which contains packages for scientific computing, **`jlab` will contain JupyterLab and the software needed to run it**. Keeping these environments separate allows us to maintain or update JupyterLab without unnecessarily changing the scientific packages used in our projects.

When we work in JupyterLab, we will be able to select a **Python kernel** associated with another environment, such as `myenv`. The kernel executes our Python code using the interpreter and packages installed in that environment. This is why we installed `ipykernel` in `myenv` in the previous section.

**Step 1: Create the JupyterLab environment**

Open your terminal and enter:

```bash
conda create -n jlab -c conda-forge --override-channels python=3.12 jupyterlab
```

Conda will identify the necessary dependencies and display the packages it proposes to install. Review the installation and enter `y` to proceed.

**Step 2: Activate the environment**

Once the installation is complete, enter:

```bash
conda activate jlab
```

Your terminal prompt should now begin with `(jlab)`.

**Step 3: Verify the JupyterLab installation**

Enter:

```bash
jupyter lab --version
```

The terminal should display the installed version of JupyterLab, confirming that it is available in the active environment.

We now have two environments with different purposes:

- **`myenv`:** Contains Python, NumPy, Matplotlib, and `ipykernel` for executing scientific Python code.
- **`jlab`:** Contains JupyterLab and its supporting software for working with notebooks.

In the next notebook, we will launch JupyterLab and configure it to use the Python kernel from `myenv`. This will allow us to work with scientific Python packages through JupyterLab while maintaining the two environments independently.

## 1.3.7 Working with pip, venv, and Conda

Earlier in this notebook, we introduced three tools for managing Python packages and environments: `pip`, `venv`, and `conda`. We have now used Conda to create and manage our own Python environments, but you will frequently encounter scientific software that provides installation instructions using `pip` and `venv` instead. For example, when downloading scientific software or following instructions from GitHub or PyPI, you may encounter commands such as:

```bash
python -m venv .venv
pip install numpy
```

The first command creates a virtual environment using Python's built-in `venv` module. The name `.venv` is simply a common convention for the directory containing that environment. The second command installs NumPy into the Python environment associated with the `pip` command. These are widely used approaches to managing Python software, and **they are not inherently better or worse than Conda**. They represent different tools and workflows for accomplishing many of the same tasks.

### Installing Packages with pip Inside a Conda Environment

An important point is that **we can use `pip` within a Conda environment**. We do not necessarily need to create a new environment using `venv` simply because software documentation provides `pip` installation instructions. For example, suppose we are working in our `myenv` environment and encounter a scientific package that is available on PyPI but not on conda-forge. Before using `pip`, we need to ensure that it is installed in the active environment. We can check by entering:

```bash
python -m pip --version
```

If `pip` is installed, this command displays its version and installation location. If Python reports that the `pip` module cannot be found, we can install it using Conda:

```bash
conda install -n myenv -c conda-forge --override-channels pip
```

Once `pip` is available, we can install packages from PyPI into our active environment:

```bash
python -m pip install package_name
```

Here, `package_name` represents the name of the package we wish to install. Using `python -m pip` explicitly runs `pip` with the Python interpreter associated with the active environment, helping ensure that the package is installed in the intended location.

However, **Conda and pip manage dependencies differently**. Conda does not fully coordinate with changes made by `pip`, and subsequent Conda operations may introduce conflicts with pip-installed packages.

For this reason, we will generally follow these practices:

1. **Look for the package on conda-forge first.** If it is available, install it using Conda.
2. **Use pip when necessary.** If the package is unavailable through Conda, we can often install it using `pip` within an existing Conda environment.
3. **Install Conda packages before pip packages.** Avoid making further Conda modifications after installing packages with `pip`, when possible.
4. **Create a new environment if necessary.** If dependencies become incompatible, it may be easier and more reliable to create a fresh environment than to repair a heavily modified one.

### When Would We Use venv Instead of Conda?

Python's built-in `venv` module provides a lightweight way to create isolated Python environments without requiring Conda. A common workflow is to create a virtual environment using `venv`, activate it, and then install the required Python packages using `pip`. This approach is widely used in Python software development, particularly for projects whose dependencies are available through PyPI and do not require extensive management of non-Python software libraries. Conda provides additional capabilities, including managing Python versions and non-Python dependencies, which makes it particularly useful for many scientific computing applications.

**In this course, we will use Conda as our primary environment manager**, but you should understand that many scientific Python projects use `venv` and `pip`. Later in the course, we will create a virtual environment using `venv` so you can compare the two approaches directly.

### Choosing the Right Tools

As you begin working with scientific software from different sources, remember that installation instructions are not necessarily tied to a single environment-management system.

A package whose documentation says `pip install` may also be available through conda-forge. Conversely, some packages may only be available through PyPI or may require specialized installation procedures. The important skill is not memorizing one set of installation commands, but **understanding which Python environment you are modifying, where packages come from, and how their dependencies are managed**.
